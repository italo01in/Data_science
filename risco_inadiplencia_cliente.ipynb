#  Data Science - Análise Preditiva 


### **Projeto:** Desenvolvimento de Modelo Preditivo para Risco de Inadimplência de Clientes.

- **Objetivo de Negócio:** Criar um modelo de classificação capaz de prever a probabilidade de um cliente se tornar inadimplente, utilizando como base seus dados de idade e renda. A finalidade é identificar clientes de alto risco para direcionar ações estratégicas e preventivas.
**Fonte de Dados:** Foi utilizado um conjunto de dados interno , contendo  registros de clientes com as seguintes informações: idade, renda e status de inadimplência (sim/não).

**Metodologia Aplicada:**

1. **Análise Exploratória:** Investigação da distribuição dos dados e da correlação entre idade, renda e o status de inadimplência por meio de estatísticas descritivas e visualização gráfica.
2. **Modelagem Preditiva:** Aplicação de um modelo de Machine Learning (Regressão Logística) para treinar o sistema a reconhecer padrões que levam à inadimplência.
3. **Validação do Modelo:** Avaliação da eficácia e precisão do modelo por meio de métricas técnicas, como Acurácia, Matriz de Confusão e Relatório de Classificação, garantindo a confiabilidade dos resultados.

---


**Fluxo de trabalho para análise preditiva**
1. Carregar os dados
2. Explorar informações 
3. Treinar um modelo simples 
4. Avaliar resultados  

### 1. Importa bibliotecas 

```python
# Bibliotescas principais

import pandas as pd  #manipular e analisar dados em Pytho
import numpy as np   # biblioteca de cálculo numérico. É usado principalmente para trabalhar com arrays e matrizes.
import matplotlib.pyplot as plt # visualização de dados
import seaborn as sns # visualização de dados mais avançados 


# Machine learning 

from sklearn.model_selection import train_test_split # Essa função é usada para dividir os dados em conjuntos de treinamento e teste.
from sklearn.linear_model import LogisticRegression #Usado para classificação binária (regressão logística).
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report  # Essas funções são usadas para avaliar o desempenho de modelos de aprendimento de máquina.
```

### 2.Carregando dados 

```python
# Dataset 

data = pd.DataFrame({
    'idade': [25, 32, 47, 51, 62, 23, 36, 45, 52, 48],
    'renda': [50000, 60000, 80000, 90000, 120000, 45000, 65000, 85000, 95000, 78000],
    'inadiplente': [0, 0, 1, 1, 1, 0, 0, 1, 1, 1]
})

data.head()
```

### 3. Exploração dos dados 

```python
# Estatisticas descritivas
print(data.describe())

# Criação de uma figura com tamanho personalizado
fig = plt.figure(figsize=(12, 8))

#Visualização dos dados
sns.scatterplot(x='idade', y='renda', hue='inadiplente', data=data)
plt.title("Distribuição de cliente por idade de renda")

plt.show()
```

### 4. Modelagem preditiva

```python
# Separando as features e o target
X = data[['idade', 'renda']]    
y = data['inadiplente']

# Dividindo os dados em conjuntos de treinamento e teste
x_train, x_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# modelo de regressão logística
model = LogisticRegression(max_iter=200)
model.fit(x_train, y_train)

# Predições
y_pred = model.predict(x_test)
```

### Avaliação do modelo

```python
# Avaliação do modelo
print("Acurácia:", accuracy_score(y_test, y_pred))
print("Matriz de Confusão:\n", confusion_matrix(y_test, y_pred))
print("Relatório de Classificação:\n", classification_report(y_test, y_pred))
```

### Conclusão
Este fluxo mostra como executo minhas análises:
- Preparação dos dados
- Exploração visual e estatística
- Modelagem preditiva com machine learning
- Avaliação de resultados
